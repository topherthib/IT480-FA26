## Setup — Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
```

```python
def plot_history(history, title, show_accuracy=False):
    if show_accuracy:
        fig, axes = plt.subplots(1, 2, figsize=(11, 4))

        axes[0].plot(history.history["loss"], label="Training Loss")
        axes[0].plot(history.history["val_loss"], label="Validation Loss")
        axes[0].set_xlabel("Epoch")
        axes[0].set_ylabel("Loss")
        axes[0].legend()
        axes[0].set_title(title + " -- Loss")

        axes[1].plot(history.history["accuracy"], label="Training Accuracy")
        axes[1].plot(history.history["val_accuracy"], label="Validation Accuracy")
        axes[1].set_xlabel("Epoch")
        axes[1].set_ylabel("Accuracy")
        axes[1].legend()
        axes[1].set_title(title + " -- Accuracy")

        plt.tight_layout()
        plt.show()
    else:
        plt.figure(figsize=(6, 4))
        plt.plot(history.history["loss"], label="Training Loss")
        plt.plot(history.history["val_loss"], label="Validation Loss")
        plt.xlabel("Epoch")
        plt.ylabel("Loss")
        plt.title(title)
        plt.legend()
        plt.show()
```

The original Wine Quality scores are imbalanced, particularly at the
extreme scores. As in Lab 1, the scores are grouped into three broader
quality classes:

- **Low (0):** quality ≤ 5
- **Medium (1):** quality = 6
- **High (2):** quality ≥ 7

```python
wine = fetch_openml(
    name="wine-quality-red",
    version=1,
    as_frame=True,
    parser="auto"
)

df3 = wine.frame

print("Dataset shape:", df3.shape)
print("\nOriginal quality distribution:")
print(df3["class"].astype(int).value_counts().sort_index())
```

```python
# Bucket raw quality scores into 3 bands: Low / Medium / High
#The resulting dataset contains three classes.

#The Low and Medium classes
#are relatively similar in size, while the High class contains substantially
#fewer examples.

def bucket_quality(q):
    q = int(q)

    if q <= 5:
        return 0  # Low
    elif q == 6:
        return 1  # Medium
    else:
        return 2  # High


df3["quality_band"] = df3["class"].apply(bucket_quality)

print("Class distribution:")
print(df3["quality_band"].value_counts().sort_index())
```

```python
X3 = df3.drop(columns=["class", "quality_band"]).values.astype(float)
y3 = df3["quality_band"].values.astype(int)

X3_train, X3_test, y3_train, y3_test = train_test_split(
    X3,
    y3,
    test_size=0.2,
    random_state=42,
    stratify=y3
)

x3_scaler = StandardScaler()

X3_train_scaled = x3_scaler.fit_transform(X3_train)
X3_test_scaled = x3_scaler.transform(X3_test)

print("Training samples:", X3_train_scaled.shape[0])
print("Testing samples:", X3_test_scaled.shape[0])
print("Number of features:", X3_train_scaled.shape[1])
```

```python
#The following model is the original multiclass MLP used in Lab 1.
#It contains two hidden layers with 128 and 32 neurons using ReLU
#activation. The output layer contains one neuron for each of the three
#classes and uses softmax activation.

n_classes = len(np.unique(y3))

model_multi = tf.keras.Sequential([
    tf.keras.layers.Dense(
        128,
        activation="relu",
        input_shape=(X3_train_scaled.shape[1],)
    ),

    tf.keras.layers.Dense(
        32,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        n_classes,
        activation="softmax"
    )
])

model_multi.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model_multi.summary()
```

```python
#The baseline model is trained using the same settings as Lab 1: 100
#epochs, a batch size of 32, and 20% of the training data reserved for
#validation.

history_multi = model_multi.fit(
    X3_train_scaled,
    y3_train,
    validation_split=0.2,
    epochs=100,
    batch_size=32,
    verbose=0
)

plot_history(
    history_multi,
    "Part A -- Original Lab 1 Baseline",
    show_accuracy=True
)
```

### Baseline Overfitting Analysis

The original Lab 1 model shows evidence of **overfitting**. Early in
training, the training and validation loss decrease together. As training
continues, however, the training loss continues to decrease while the
validation loss begins to increase substantially.

The growing gap between training and validation loss indicates that the
model is continuing to improve its fit to the training data without
improving its ability to generalize to unseen validation data.

### Test Performance

```python
multi_probs = model_multi.predict(X3_test_scaled)
multi_preds = np.argmax(multi_probs, axis=1)

acc_multi = accuracy_score(y3_test, multi_preds)

print(f"Baseline Test Accuracy: {acc_multi:.3f}")

cm3 = confusion_matrix(y3_test, multi_preds)

ConfusionMatrixDisplay(
    cm3,
    display_labels=["Low", "Medium", "High"]
).plot()

plt.title("Part A -- Baseline Confusion Matrix")
plt.show()
```

### Baseline Results

The original Lab 1 run achieved a test accuracy of **66.2%**.

The confusion matrix showed that the model most frequently confused the
**Low and Medium** classes. Medium and High wines were also confused, but
less frequently. This is reasonable because the classes were created by
grouping neighboring numerical wine-quality scores into broader categories.

Overall, the baseline model achieved reasonable classification accuracy,
but the training and validation loss curves showed that the model was
overfitting as training continued.

# Part B — Hyperparameter Tuning

```python
!pip install keras-tuner --quiet
import keras_tuner as kt

n_features = X3_train_scaled.shape[1]
n_classes = 3

def build_model(hp):
    model = tf.keras.Sequential([
        tf.keras.layers.Dense(
            hp.Int("hidden1", min_value=16, max_value=128, step=16),
            activation="relu",
            input_shape=(n_features,)
        ),

        tf.keras.layers.Dense(
            hp.Int("hidden2", min_value=8, max_value=64, step=8),
            activation="relu"
        ),

        tf.keras.layers.Dense(
            n_classes,
            activation="softmax"
        )
    ])

    learning_rate = hp.Choice(
        "learning_rate",
        values=[0.0001, 0.001, 0.01]
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        ),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

```python
tuner = kt.RandomSearch(
    build_model,
    objective="val_loss",
    max_trials=10,
    overwrite=True,
    directory="tuning",
    project_name="wine_quality"
)

tuner.search(
    X3_train_scaled,
    y3_train,
    validation_split=0.2,
    epochs=50,
    verbose=1
)
```

```python
best_hp = tuner.get_best_hyperparameters(num_trials=1)[0]

print("Best hyperparameters:")
print("Hidden layer 1:", best_hp.get("hidden1"))
print("Hidden layer 2:", best_hp.get("hidden2"))
print("Learning rate:", best_hp.get("learning_rate"))
```

```python
tuned_model = tuner.hypermodel.build(best_hp)

history_tuned = tuned_model.fit(
    X3_train_scaled,
    y3_train,
    validation_split=0.2,
    epochs=50,
    verbose=0
)

test_loss, test_accuracy = tuned_model.evaluate(
    X3_test_scaled,
    y3_test,
    verbose=0
)

print(f"Tuned Test Accuracy: {test_accuracy:.3f}")
```

### Confirming Changes in Accuracy

```python
improvement = test_accuracy - acc_multi

print(f"Baseline Accuracy: {acc_multi:.3f}")
print(f"Tuned Accuracy:    {test_accuracy:.3f}")
print(f"Change:            {improvement:+.3f}")
```

# Part C — Addressing Overfitting


```python
regularized_model = tf.keras.Sequential([
    tf.keras.layers.Dense(
        best_hp.get("hidden1"),
        activation="relu",
        input_shape=(n_features,)
    ),

    tf.keras.layers.Dropout(0.3),

    tf.keras.layers.Dense(
        best_hp.get("hidden2"),
        activation="relu"
    ),

    tf.keras.layers.Dropout(0.3),

    tf.keras.layers.Dense(
        n_classes,
        activation="softmax"
    )
])

regularized_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=best_hp.get("learning_rate")
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=10,
    restore_best_weights=True
)
```

```python
history_regularized = regularized_model.fit(
    X3_train_scaled,
    y3_train,
    validation_split=0.2,
    epochs=200,
    callbacks=[early_stop],
    verbose=0
)
```

### Display New Findings

```python
# Plot Part C training and validation loss/accuracy
plot_history(
    history_regularized,
    "Part C -- Regularized Tuned Model",
    show_accuracy=True
)

# Direct comparison of Part A baseline and Part C regularized model

plt.figure(figsize=(10, 6))

# Part A
plt.plot(
    history_multi.history["loss"],
    label="Part A - Training Loss"
)

plt.plot(
    history_multi.history["val_loss"],
    label="Part A - Validation Loss"
)

# Part C
plt.plot(
    history_regularized.history["loss"],
    label="Part C - Training Loss"
)

plt.plot(
    history_regularized.history["val_loss"],
    label="Part C - Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Part A vs. Part C -- Training and Validation Loss")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
# Final Part C test evaluation

test_loss_c, test_accuracy_c = regularized_model.evaluate(
    X3_test_scaled,
    y3_test,
    verbose=0
)

multi_probs_c = regularized_model.predict(
    X3_test_scaled,
    verbose=0
)

multi_preds_c = np.argmax(multi_probs_c, axis=1)

print(f"Part A Baseline Test Accuracy: {acc_multi:.3f}")
print(f"Part C Final Test Accuracy:    {test_accuracy_c:.3f}")
print(f"Accuracy Change:               {test_accuracy_c - acc_multi:+.3f}")
```

```python
# Direct comparison of Part A and Part C confusion matrices

cm_a = confusion_matrix(y3_test, multi_preds)
cm_c = confusion_matrix(y3_test, multi_preds_c)

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

ConfusionMatrixDisplay(
    confusion_matrix=cm_a,
    display_labels=["Low", "Medium", "High"]
).plot(
    ax=axes[0],
    colorbar=False
)

axes[0].set_title("Part A -- Baseline")

ConfusionMatrixDisplay(
    confusion_matrix=cm_c,
    display_labels=["Low", "Medium", "High"]
).plot(
    ax=axes[1],
    colorbar=False
)

axes[1].set_title("Part C -- Regularized")

plt.tight_layout()
plt.show()
```

### Results and Comparison

The Part C model achieved a final test accuracy of **64.7%**, compared
with the Part A baseline accuracy of **68.8%** in the current run.

Compared with the original Part A loss curve, the gap between training
and validation loss **decreased substantially**. In Part A, training loss
continued to decrease while validation loss increased as training
continued, showing clear overfitting. In Part C, the training and
validation loss remained much closer together. Early Stopping ended
training after approximately **16 epochs**, preventing the model from
continuing to train after validation loss stopped improving.

The confusion matrix shows that the Part C model most frequently confused
the **Medium and Low** classes and the **High and Medium** classes.
Compared with Part A, Low-to-Medium errors decreased from 36 to 23, but
Medium-to-Low errors increased from 31 to 50. The largest change occurred
for the High class, where High-to-Medium errors increased from 15 to 38.

Overall, Dropout and Early Stopping substantially reduced the overfitting
seen in Part A, but this did not improve the final test accuracy. The
regularized model became better at identifying Low-quality samples but
worse at distinguishing the High-quality class.

# Part D — Reflection

It is important to use the test set only at the end because it provides an unbiased measure of how well the final model performs on data it has never seen during training or model selection. If I had used the test set to select the best hyperparameter configuration, I would have indirectly tuned the model to the test data, making the final test accuracy less representative of its ability to generalize to new data. Dropout and Early Stopping meaningfully reduced the overfitting observed in Part A, as the Part C training and validation loss curves remained much closer together and Early Stopping ended training after approximately 16 epochs instead of allowing training to continue for all 100 epochs. However, reducing overfitting did not improve overall test performance, as accuracy decreased from approximately 68.8% in the Part A run to 64.7% in Part C, with the regularized model particularly struggling to correctly classify the High-quality class. If I had unlimited time and computing resources, I would expand the hyperparameter search to include additional layer sizes, learning rates, Dropout rates, batch sizes, and network depths, while also investigating methods for addressing the class imbalance that may be contributing to the poor performance on the High-quality class.
